```python
# Program: Catatan Diet Giant
# Disusun oleh: [Alif fathan fatazka ramadhan]
# Tujuan: Membantu Giant mencatat data diet harian

import math

# 1. Data Diri Giant
nama = "Giant"
usia = 20
berat_badan = 97.5  # kg
tinggi_badan = 1.75  # meter
status_olahraga = False
```

```python
# 2. Konversi berat dari string ke float dan buat kalimat pengantar
berat_str = "95.7"
berat_float = float(berat_str)

print(f"Halo, saya {nama}. Usia saya {usia} tahun dan berat saya {berat_float} kg")
```

```python
# 3. Hitung BMI: BMI = berat / (tinggi ** 2)
bmi = berat_badan / (tinggi_badan ** 2)
bmi_bulat = round(bmi, 2)

print(f"Nilai BMI asli: {bmi}")
print(f"Nilai BMI dibulatkan 2 desimal: {bmi_bulat}")
```

```python
# 4. Target minum air
target_air = 2000  # ml
isi_botol = 350    # ml per botol

hasil_asli = target_air / isi_botol         # true division
botol_penuh = target_air // isi_botol       # floor division
sisa_air = target_air % isi_botol           # modulo

print(f"Hasil pembagian asli: {hasil_asli} botol")
print(f"Botol penuh: {botol_penuh} botol")
print(f"Sisa air: {sisa_air} ml")
```

```python
# 5. Kalori harian
kalori_harian = 2100.5
kalori_floor = math.floor(kalori_harian)
kalori_ceil = math.ceil(kalori_harian)

print(f"Kalori asli: {kalori_harian}")
print(f"Kalori dibulatkan ke bawah: {kalori_floor}")
print(f"Kalori dibulatkan ke atas: {kalori_ceil}")
```
